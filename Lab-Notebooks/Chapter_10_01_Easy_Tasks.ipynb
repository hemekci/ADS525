### Model Loading

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
# Load a pre-trained embedding model
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

### Helper Functions

```python
def compute_similarity(sentence1, sentence2):
    """Compute cosine similarity between two sentences"""
    emb1 = model.encode([sentence1])
    emb2 = model.encode([sentence2])
    return cosine_similarity(emb1, emb2)[0][0]
```

Calculate similarity for different pairs.

```python
# Pair 1: Cat/feline (similar meaning, different words)
sent1_p1 = "The cat is sleeping on the couch."
sent2_p1 = "A feline is resting on the sofa."

score_p1 = compute_similarity(sent1_p1, sent2_p1)
print(f"Pair 1: {sent1_p1[:40]}... <-> {sent2_p1[:40]}...")
print(f"Similarity: {score_p1:.4f}")
```

```python
# Pair 2: Cat/basketball (completely different topics)
sent1_p2 = "The cat is sleeping on the couch."
sent2_p2 = "I enjoy playing basketball."

score_p2 = compute_similarity(sent1_p2, sent2_p2)
print(f"\nPair 2: {sent1_p2[:40]}... <-> {sent2_p2[:40]}...")
print(f"Similarity: {score_p2:.4f}")
```

```python
# Pair 3: ML/AI (similar domain)
sent1_p3 = "Machine learning is fascinating."
sent2_p3 = "AI and ML are interesting topics."

score_p3 = compute_similarity(sent1_p3, sent2_p3)
print(f"\nPair 3: {sent1_p3[:40]}... <-> {sent2_p3[:40]}...")
print(f"Similarity: {score_p3:.4f}")
```

```python
# Pair 4: Pizza (same meaning, different structure)
sent1_p4 = "I love pizza."
sent2_p4 = "Pizza is my favorite food."

score_p4 = compute_similarity(sent1_p4, sent2_p4)
print(f"\nPair 4: {sent1_p4[:40]}... <-> {sent2_p4[:40]}...")
print(f"Similarity: {score_p4:.4f}")
```

```python
# Pair 5: Weather (opposite conditions)
sent1_p5 = "The weather is nice today."
sent2_p5 = "It's raining heavily outside."

score_p5 = compute_similarity(sent1_p5, sent2_p5)
print(f"\nPair 5: {sent1_p5[:40]}... <-> {sent2_p5[:40]}...")
print(f"Similarity: {score_p5:.4f}")
```

### Task 1a: Find Most and Least Similar Pairs

Based on the scores above, identify which pair is most similar and which is least similar.

```python
# Collect all scores
all_scores = [score_p1, score_p2, score_p3, score_p4, score_p5]
all_pairs = [
    (sent1_p1, sent2_p1),
    (sent1_p2, sent2_p2),
    (sent1_p3, sent2_p3),
    (sent1_p4, sent2_p4),
    (sent1_p5, sent2_p5)
]

# Find extremes
most_similar_idx = np.argmax(all_scores)
least_similar_idx = np.argmin(all_scores)

print("Most similar pair:")
print(f"  {all_pairs[most_similar_idx][0]}")
print(f"  {all_pairs[most_similar_idx][1]}")
print(f"  Similarity: {all_scores[most_similar_idx]:.4f}")
```

```python
print("Least similar pair:")
print(f"  {all_pairs[least_similar_idx][0]}")
print(f"  {all_pairs[least_similar_idx][1]}")
print(f"  Similarity: {all_scores[least_similar_idx]:.4f}")
```

#### Easy Task 2: Comparing Loss Functions

### Instructions

We'll compare the default model with a base model (minimal training).

```python
# Load different models
trained_model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
base_model = SentenceTransformer('bert-base-uncased')
```

```python
def compare_models(sent1, sent2, model1, model2, name1="Model 1", name2="Model 2"):
    """Compare similarity scores from two different models"""
    # Model 1
    emb1_m1 = model1.encode([sent1])
    emb2_m1 = model1.encode([sent2])
    sim_m1 = cosine_similarity(emb1_m1, emb2_m1)[0][0]
    
    # Model 2
    emb1_m2 = model2.encode([sent1])
    emb2_m2 = model2.encode([sent2])
    sim_m2 = cosine_similarity(emb1_m2, emb2_m2)[0][0]
    
    print(f"\nSentence 1: {sent1}")
    print(f"Sentence 2: {sent2}")
    print(f"{name1} similarity: {sim_m1:.4f}")
    print(f"{name2} similarity: {sim_m2:.4f}")
    print(f"Difference: {abs(sim_m1 - sim_m2):.4f}")
    
    return sim_m1, sim_m2
```

Compare on semantic pair 1.

```python
# Pair 1: Dog/puppy
sem_sent1_p1 = "The dog is playing in the garden."
sem_sent2_p1 = "A puppy is having fun in the yard."

sim_trained_p1, sim_base_p1 = compare_models(
    sem_sent1_p1, sem_sent2_p1, trained_model, base_model, "Trained", "Base"
)
```

Compare on semantic pair 2.

```python
# Pair 2: Groceries/shopping
sem_sent1_p2 = "I need to buy groceries."
sem_sent2_p2 = "I should go shopping for food."

sim_trained_p2, sim_base_p2 = compare_models(
    sem_sent1_p2, sem_sent2_p2, trained_model, base_model, "Trained", "Base"
)
```

Compare on semantic pair 3.

```python
# Pair 3: Boring/uninteresting
sem_sent1_p3 = "The movie was boring."
sem_sent2_p3 = "The film was uninteresting."

sim_trained_p3, sim_base_p3 = compare_models(
    sem_sent1_p3, sem_sent2_p3, trained_model, base_model, "Trained", "Base"
)
```

#### Easy Task 3: Preparing Training Data

### Instructions

1. Load a dataset with sentence pairs
2. Transform data for SoftmaxLoss (classification)
3. Transform data for CosineSimilarityLoss (similarity)
4. Transform data for MultipleNegativesRankingLoss (ranking)
5. Verify data format for each loss function

```python
from datasets import load_dataset, Dataset

# Load a small subset of MNLI
raw_data = load_dataset("glue", "mnli", split="train").select(range(100))
```

```python
# View raw data structure
print("Raw data example:")
print(raw_data[0])
print(f"\nLabel meanings: 0=entailment, 1=neutral, 2=contradiction")
```

### Format 1: SoftmaxLoss (Classification)

SoftmaxLoss needs premise, hypothesis, and a discrete label.

```python
# This format is already correct for SoftmaxLoss
softmax_data = raw_data.remove_columns("idx")

print("SoftmaxLoss format:")
print(softmax_data[0])
print(f"\nFeatures: {softmax_data.column_names}")
```

### Format 2: CosineSimilarityLoss (Similarity)

CosineSimilarityLoss needs sentence1, sentence2, and a continuous label (0-1).

```python
# Transform: entailment=1, neutral/contradiction=0
mapping = {0: 1.0, 1: 0.0, 2: 0.0}

cosine_data = Dataset.from_dict({
    "sentence1": raw_data["premise"],
    "sentence2": raw_data["hypothesis"],
    "label": [mapping[label] for label in raw_data["label"]]
})

print("CosineSimilarityLoss format:")
print(cosine_data[0])
print(f"\nFeatures: {cosine_data.column_names}")
```

### Format 3: MultipleNegativesRankingLoss (Ranking)

MNRLoss needs anchor, positive, and optionally negative sentences.

```python
# Filter only entailment pairs (label=0)
entailment_data = raw_data.filter(lambda x: x['label'] == 0)

mnr_data = Dataset.from_dict({
    "anchor": entailment_data["premise"],
    "positive": entailment_data["hypothesis"],
})

print("MultipleNegativesRankingLoss format:")
print(mnr_data[0])
print(f"\nFeatures: {mnr_data.column_names}")
print(f"Number of pairs: {len(mnr_data)}")
```

### Task 3a: Add Negatives to MNR Data

Your task: Add a negative example to each anchor-positive pair.

```python
import random

# Fill in: Add negative examples
# Hint: Use contradiction pairs (label=2) or random shuffle

negatives = list(entailment_data["hypothesis"])
random.shuffle(negatives)

mnr_with_negatives = Dataset.from_dict({
    "anchor": entailment_data["premise"],
    "positive": entailment_data["hypothesis"],
    "negative": negatives
})

print("MNR with negatives:")
print(mnr_with_negatives[0])
```

#### Easy Task 4: Using Similarity Evaluators

### Instructions

1. Load evaluation dataset (STS-B)
2. Create EmbeddingSimilarityEvaluator
3. Evaluate a pre-trained model
4. Compare different similarity metrics
5. Interpret evaluation scores

```python
from sentence_transformers.evaluation import EmbeddingSimilarityEvaluator

# Load STS-B validation set
val_sts = load_dataset('glue', 'stsb', split='validation')
```

```python
# View evaluation data
print("STS-B example:")
print(val_sts[0])
print(f"\nLabel range: 0 (no similarity) to 5 (high similarity)")
```

Create an evaluator with cosine similarity.

```python
# Create evaluator
evaluator = EmbeddingSimilarityEvaluator(
    sentences1=val_sts["sentence1"],
    sentences2=val_sts["sentence2"],
    scores=[score/5 for score in val_sts["label"]],  # Normalize to 0-1
    main_similarity="cosine"
)

print("Evaluator created")
print(f"Number of test pairs: {len(val_sts)}")
```

Evaluate the model.

```python
# Evaluate trained model
results = evaluator(trained_model)

print("\nEvaluation Results:")
for metric, score in results.items():
    print(f"{metric}: {score:.4f}")
```

### Task 4a: Compare Different Models

Evaluate the base model and compare with the trained model.

```python
# Evaluate base model
base_results = evaluator(base_model)

print("\nBase Model Results:")
for metric, score in base_results.items():
    print(f"{metric}: {score:.4f}")
```

Compare Pearson correlation.

```python
print("\nPearson Correlation Comparison:")
print(f"Trained - Pearson Cosine: {results['pearson_cosine']:.4f}")
print(f"Base    - Pearson Cosine: {base_results['pearson_cosine']:.4f}")
print(f"Improvement: {results['pearson_cosine'] - base_results['pearson_cosine']:.4f}")
```

Compare Spearman correlation.

```python
print("\nSpearman Correlation Comparison:")
print(f"Trained - Spearman Cosine: {results['spearman_cosine']:.4f}")
print(f"Base    - Spearman Cosine: {base_results['spearman_cosine']:.4f}")
print(f"Improvement: {results['spearman_cosine'] - base_results['spearman_cosine']:.4f}")
```
